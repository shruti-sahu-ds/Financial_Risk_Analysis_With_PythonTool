# Financial Risk Analysis with Python  
## Task 5: Visualisation   

**Objective:**  
To perform exploratory data analysis (EDA) and present key transaction and risk insights using clear and informative visualizations that support financial behavior and risk assessment.


```python
# Extensive Exploratory Data Analysis (EDA)

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load cleaned data
df = pd.read_csv("cleaned_transactions.csv")
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])
```

```python
# --- 1. Transaction Amount Distribution ---
plt.figure(figsize=(8, 5))
plt.hist(df['TransactionAmount'], bins=30)
plt.title('Distribution of Transaction Amounts')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# --- 2. Account Balance Distribution ---
plt.figure(figsize=(8, 5))
plt.hist(df['AccountBalance'], bins=30)
plt.title('Distribution of Account Balances')
plt.xlabel('Account Balance')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# --- 3. Transaction Type Count ---
txn_type_counts = df['TransactionType'].value_counts()

plt.figure(figsize=(7, 5))
txn_type_counts.plot(kind='bar')
plt.title('Transaction Count by Transaction Type')
plt.xlabel('Transaction Type')
plt.ylabel('Number of Transactions')
plt.grid(axis='y')
plt.show()
```

```python
# --- 4. Average Balance by Account Type ---
avg_balance_by_account = (
    df.groupby('AccountType')['AccountBalance']
      .mean()
      .sort_values()
)
```

```python
plt.figure(figsize=(7, 5))
avg_balance_by_account.plot(kind='bar')
plt.title('Average Account Balance by Account Type')
plt.xlabel('Account Type')
plt.ylabel('Average Balance')
plt.grid(axis='y')
plt.show()
```

```python
# 5. Boxplot for Transaction Amounts (Outliers)
plt.figure(figsize=(8, 4))
plt.boxplot(df['TransactionAmount'], vert=False)
plt.title('Transaction Amounts – Outlier Detection')
plt.xlabel('Transaction Amount')
plt.show()
```
