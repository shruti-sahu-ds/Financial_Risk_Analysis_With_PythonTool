**Role:** Financial Data Analyst – Morgan Stanley  
**Objective:** Clean and standardize customer transaction data to ensure accuracy for downstream analysis.



```python
import numpy as np
import pandas as pd
import warnings 
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r"C:\Users\HP PAVILLION\Downloads\morgan_stanely.csv")
```

```python
df.info()
```

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Remove special characters and non-numeric entries from financial fields
financial_cols = ['TransactionAmount', 'AccountBalance']

for col in financial_cols:
    df[col] = (
        df[col]
        .astype(str)
        .replace('[^0-9.-]', '', regex=True)
        .astype(float)
    )

```

```python
# Convert currency amounts into numerical format
financial_cols = ['TransactionAmount', 'AccountBalance']

for col in financial_cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).replace('[^0-9.-]', '', regex=True),
        errors='coerce'
    )
```

```python
# Validate and format date columns
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'], errors='coerce')
```

```python
# Ensure account types and transaction categories are standardized
df['AccountType'] = df['AccountType'].str.lower().str.strip()
df['TransactionType'] = df['TransactionType'].str.lower().str.strip()
```

```python
# Save cleaned data for use in Task 2
df.to_csv("cleaned_transactions.csv", index=False)
```
